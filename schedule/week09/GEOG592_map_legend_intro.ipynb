```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python


# getting the US census data

url_0 = "https://www2.census.gov/geo/tiger/GENZ2024/shp/" # url for census tiger data 
url_county = "cb_2024_us_county_500k.zip" # county level
url_state = "cb_2024_us_state_500k.zip" # state level 


## Code for county level data 

counties = gpd.read_file(url_0+url_county)
nc_counties = counties.loc[counties["STATEFP"] == "37"].copy()
nc_counties.plot(edgecolor="black", facecolor="white")
```

```python
ncc = nc_counties.to_crs(epsg = 6543).copy()
ncc.plot(edgecolor="black")
plt.title("North Carolina Counties")
```

```python
def ncplot(ax, stroke, fill):
    ncc.plot(ax=ax, edgecolor=stroke, facecolor=fill)

fig, ax = plt.subplots(figsize=(20,10))
ncplot(ax, "black", "gray")
plt.show
             
```

```python
ncc_copy = ncc.copy()
ncc_copy["geometry"] = ncc_copy.geometry.centroid
ncc_copy[ncc_copy["NAME"] == "Durham"]
```

Assign each of the counties labels on the map (TINY font size) use loop to assign x and y point of centroid and assign the name



```python
ncc_copy = ncc.copy()
ncc_copy["geometry"] = ncc_copy.geometry.centroid

fig, ax = plt.subplots(figsize=(20,10))
ncplot(ax, "black", "gray")

for name in ncc_copy["NAME"]:
    point = ncc_copy[ncc_copy["NAME"] == name]["geometry"]
    x = point.x.iloc[0]
    y = point.y.iloc[0]
    ax.text(x, y, name, ha="center", va="center", fontsize=8, fontname="Times New Roman")

ax.set_title("North Carolina Counties", fontsize=20, pad=25, fontname="Times New Roman")

    

```
