```python
import csv

filename = "data/NC_Fire_Stations.csv"

with open(filename, "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print("Number of rows:", len(rows))
print("Number of columns:", len(reader.fieldnames))
print("Column names:")
print(reader.fieldnames)
```

Importing and exploring the structure of the CSV file I found. The file appears to contain 2126 rows, one for each fire station in North Carolina, along with corresponding information like the address and longitude and latitude (x and y coordinates). Therefore, when we clean the data, we will need to keep the x and y coordinates. 

```python
station_data_cleaned = []

def can_be_float(value):
    try:
        float(value)
        return True
    except (ValueError, TypeError):
        return False

def can_be_int(value):
    try:
        int(value)
        return True
    except ValueError:
        return False


for row in rows:
    xvalue = row["x"]
    yvalue = row["y"]
    
    if can_be_float(xvalue) and can_be_float(yvalue):
        station_data_cleaned.append({
                "name": row["DEPT_NAME"],
                "x": row["x"],
                "y": row["y"]
            })
    else:
         print(f"Error found in coordinates at {row}")
    
```

This code was used to check the x and y coordinates that would be used to spatially join this data. I used a for loop to find any longitude or latitude coordinates that were not able to be converted to a float, meaning they either were missing data or had some other error. The code found that the coordinates are missing for one fire station in Winston Salem. As both the x and y, as well as the latitude and longitude are missing, we have no data to substitute in, so we will just leave it out of the cleaned data output. I appended all float values to a new list along with their name. 

```python
error_found = False

for row in station_data_cleaned:
    value = row["name"]
    if value == "":
        print(f" {row} is missing a name")
        error_found = True

if not error_found:
    print("Everything looks good.")
    
```

This code chunk is inspecting if every row of coordinates has a name associated with it. As no errors were found, we can concluded that the name column is good!

```python
fieldnames = station_data_cleaned[0].keys() 

with open("data/station_data_cleaned.csv", "w", newline="") as file: 
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(station_data_cleaned)
```

This last bit of code exports the cleaned csv file to the desired data folder.
