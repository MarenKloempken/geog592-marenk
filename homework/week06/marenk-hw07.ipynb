11. Would EPSG:4326 be a good CRS for measuring distance in feet? Explain briefly. 

No, 4326 would be a bad choice of CRS to measure in feet, as 4326 is a coordinate system based on decimal degrees and therefore, does not measure in feet. For feet, you should use the local projection EPSG 6543.

21. Why is a spatial join different from a regular pandas join on a common ID field? 

The spatial join typically has a condition that performs the join based on certain geometric relationships between the data. A regualr pandas join does not use geometry to carry out a join.

24. Why can a polygon-to-polygon `intersects` spatial join create multiple output rows if each original layer
