```python
import csv

filename = "tabela1613.csv"

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.reader(file)

    for x in range(4):
        next(reader)

    headers = next(reader)

    # Create dictionaries using those headers
    reader = csv.DictReader(file, fieldnames=headers)

    rows = list(reader)

cleaned = (rows)
print(cleaned[:5])

```

```python
in_para = []

for row in cleaned:
    value = row["Município"]

    if value is not None and "(PA)" in value:
        in_para.append(row)
```

I had to search for "(PA)" in the municipalities values to select for all municipalities in Para. However, I got an TypeError around the none value, so I had to include that in my if loop. 

```python
selected_columns = []

for row in in_para:

    selected_columns.append({
        "Cód.": row["Cód."],
        "Açaí": row["Açaí"]
    })

for row in selected_columns[:5]:
    print(row)

print(len(selected_columns))
```

I had to create a new dictionary that combined the already filtered dictionary of municipalities in para, while now only selecting for the desired columns (cod and acai). 

```python
def can_be_int(value):
    try:
        int(value)
        return True
    except (ValueError, TypeError):
        return False


filtered_columns = []

for row in selected_columns:
    value = row["Açaí"]

    if can_be_int(value):
        filtered_columns.append(row)

selected_columns = filtered_columns
print(len(selected_columns))
print(selected_columns[:5])

```

This was added to remove all of the values in the acai production column that cannot be converted to an integer. This acts as a catch all, for it will remove any values that are missing or any that include text strings but not numbers (such as ... or "-"). 

```python
def can_be_int(value):
    try:
        int(value)
        return True
    except ValueError:
        return False


error_found = False

for row in selected_columns:
    value = row["Cód."]

    if value == "":
        print(
            row["Cód."],
            "is missing a Cód."
        )
        error_found = True

    elif not can_be_int(value):
        print(
            row["Cód."],
            "has suspicious Cód. value:",
            value
        )
        error_found = True

if not error_found:
    print("Everything looks good.")
```

This was to check if there were any errors in the cod column. There were no errors found so nothing further needed to be cleaned from the cod.

```python
fieldnames = selected_columns[0].keys()

with open("selected_columns.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(selected_columns)
```

This is the code to export the csv. Again, we have to specify utf-8, due to the special characters that are used in the file.
